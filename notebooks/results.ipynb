# Case-oriented review of the interview codebook

## Load and validate the current workbook

```python
from pathlib import Path
import sys, importlib
from IPython.display import display
import pandas as pd
PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'scripts').is_dir(): PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))
load_and_validate = importlib.import_module('scripts.01_load_and_validate').load_and_validate
validation = load_and_validate()
codes = validation['codes']
display(pd.DataFrame({'measure': ['Participants', 'Coded excerpt rows', 'Question IDs', 'Manual themes'], 'count': [codes.participant_id.nunique(), len(codes), codes.question_id.nunique(), codes.theme.nunique()]}))
display(validation['questions'])
from scripts._common import CODEBOOK_WORKBOOK
available_sheets = pd.ExcelFile(CODEBOOK_WORKBOOK).sheet_names
if 'Theme Matrix' in available_sheets: display(validation['matrix_diff'])
else: print('Theme Matrix check unavailable: the current workbook has no Theme Matrix sheet.')
if 'Overview' in available_sheets: display(validation['stats_diff'])
else: print('Overview check unavailable: the current workbook has no Overview sheet.')
```
